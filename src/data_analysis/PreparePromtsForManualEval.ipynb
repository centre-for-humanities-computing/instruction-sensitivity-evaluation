```python
import json, os
import pandas as pd
```

```python
data = json.load(open("../../data/prompts/openai-gpt-oss-120b-gen-prompts.json", 'r'))
```

```python
len(data)
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df = df[['task', 'dataset_description', 'language', 'generated_prompt']].drop_duplicates()
```

```python
df.shape
```

```python
def prepare_review_sample(row):
    return f"{row['task']}\nDesc:{row['dataset_description']}\nLang:\n{row['language']}\n\nGen Prompt:\n{row['generated_prompt']}"

df['text'] = df.apply(prepare_review_sample, axis=1)
```

```python
df
```

```python
df.to_csv("../../data/manual_eval_prompts.csv")
```
